Data Set Cleaning and uploading.

```python
import pandas as pd

# Upload file
from google.colab import files
uploaded = files.upload()

# Read dataset
df = pd.read_csv('Sales Dataset.csv')

# Standardize text columns
text_cols = ['Category','Sub-Category','CustomerName','State','City','PaymentMode']

for col in text_cols:
    df[col] = df[col].astype(str).str.strip().str.title()

# Remove duplicate rows
df = df.drop_duplicates()

# Convert date column
df['Order Date'] = pd.to_datetime(df['Order Date'])

# Create Year, Month, Quarter
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month_name()
df['Quarter'] = df['Order Date'].dt.quarter

# Check missing values
print(df.isnull().sum())

# Save cleaned file
df.to_csv('Cleaned_Sales_Dataset.csv', index=False)

# Download cleaned file
files.download('Cleaned_Sales_Dataset.csv')

print("✅ Cleaning Complete")
```
