```python
import pandas as pd

# Load the dataset
df = pd.read_csv('sales_data.csv')

# Clean column names (strip whitespace)
df.columns = df.columns.str.strip()

# Preview data
print(df.head())

# Basic summary
print(df.describe())

# Total sales
total_sales = df['Final Price'].sum()
print("Total Sales:", total_sales)

# Discount analysis
discounted_items = df[df['Discount'] > 0]
print("Items with Discounts:\n", discounted_items)

# Group by Item
sales_by_item = df.groupby('Item Details')['Final Price'].sum()
print("Sales by Item:\n", sales_by_item)

# Visualization
import matplotlib.pyplot as plt
sales_by_item.plot(kind='bar', title='Sales by Item')
plt.show()
```
