**Q1.** An order has `unit_price = 2499`, `quantity = 4`, `discount_pct = 0.15` (15%), and `gst_pct = 0.18` (18% applied after discount).
Calculate and print:
1. `gross_amount` (`unit_price * quantity`)
2. `net_amount` (after subtracting discount)
3. `final_bill` (after adding 18% GST to `net_amount`, rounded to 2 decimal places)

```python
# ✍️ Gross Amount: 9996
# Net Amount: 8496.6
# Final Bill: 10025.99
```

```python
# ✍️ Code

unit_price = 2499
quantity = 4
discount_pct = 0.15
gst_pct = 0.18

gross_amount = unit_price * quantity

net_amount = gross_amount - (gross_amount * discount_pct)

final_bill = net_amount + (net_amount * gst_pct)

print("Gross Amount:", gross_amount)
print("Net Amount:", net_amount)
print("Final Bill:", round(final_bill, 2))
```

**Q3.** A CRM export has a messy customer string:
`raw_customer = "   rohan sharma | MUMBAI | rohan.sharma@TCS.com   "`
Using string methods (`strip`, `split`, `title`, `lower`), extract and print:
* `customer_name` in Title Case (`"Rohan Sharma"`)
* `city` in Title Case (`"Mumbai"`)
* `email_domain` in lowercase (`"tcs.com"`)

**Q7.** Given daily store sales for 10 days:
`daily_sales = [42000, 55000, 38000, 61000, 49000, 72000, 31000, 58000, 64000, 45000]`
Using a `for` loop (without importing any library), calculate and print:
1. The **mean** daily sale
2. How many days beat the daily target of `₹50,000`

```python
# ✍️ Your answer
# Mean Daily Sale: 51500.0
# Days Above Target: 5
```

```python
# ✍️ Your code here
daily_sales = [42000, 55000, 38000, 61000, 49000, 72000, 31000, 58000, 64000, 45000]

total_sales = 0
target_days = 0

for sale in daily_sales:
    total_sales = total_sales + sale

    if sale > 50000:
        target_days = target_days + 1

mean_sales = total_sales / len(daily_sales)

print("Mean Daily Sale:", mean_sales)
print("Days Above Target:", target_days)
```

**Q9.** A list of order amounts contains negative numbers (returns) and zeros (cancelled orders):
`orders = [1200, 3400, 0, -500, 2800, 0, 1500, -1200, 4100, 900]`
Write a loop that skips `<= 0` values (using `continue`) and computes the **count**, **total**, and **mean** of valid positive orders only.

```python
# ✍️ Your answer
# Valid Orders Count: 6
# Total Valid Orders: 13900
# Mean Order Value: 2316.6666666666665
```

```python
# ✍️ Your code here
orders = [1200, 3400, 0, -500, 2800, 0, 1500, -1200, 4100, 900]

count = 0
total = 0

for order in orders:

    if order <= 0:
        continue

    count = count + 1
    total = total + order

mean = total / count

print("Valid Orders Count:", count)
print("Total Valid Orders:", total)
print("Mean Order Value:", mean)
```

**Q10.** A startup currently has `monthly_users = 5000` and grows by `12%` (`0.12`) every month.
Use a `while` loop to find how many months it takes for `monthly_users` to cross `10000`. Print the month count and final user count.

```python
# ✍️ Your answer
# Months Required: 7
# Final User Count: 11053.41
```

```python
# ✍️ Your code here
monthly_users = 5000
growth_rate = 0.12
months = 0

while monthly_users <= 10000:
    monthly_users = monthly_users * (1 + growth_rate)
    months = months + 1

print("Months Required:", months)
print("Final User Count:", round(monthly_users, 2))
```

**Q11.** Write a function `calc_profit_margin(revenue, cost)` that returns the profit margin percentage `((revenue - cost) / revenue) * 100` rounded to 2 decimals. If `revenue == 0`, return `0.0`.
Test it on `(50000, 36000)` and `(0, 500)`.

```python
# ✍️ Your answer
28.0
0.0
# meaning= The company earns ₹28 profit for every ₹100 of revenue generated.
```

```python
# ✍️ Your code here
def calc_profit_margin(revenue, cost):

    if revenue == 0:
        return 0.0

    profit_margin = ((revenue - cost) / revenue) * 100

    return round(profit_margin, 2)


# Test Cases

print(calc_profit_margin(50000, 36000))
print(calc_profit_margin(0, 500))
```

**Q12.** Write a function `net_price(mrp, discount=0.10, tax=0.18)` with default arguments (`10%` discount, `18%` tax).
Call it three ways and print the results:
1. `net_price(1000)` (uses both defaults)
2. `net_price(1000, discount=0.20)` (20% discount, default tax)
3. `net_price(1000, discount=0.0, tax=0.05)` (no discount, 5% tax)

```python
# ✍️ Your answer
1062.0
944.0
1050.0
```

```python
# ✍️ Your code here
def net_price(mrp, discount=0.10, tax=0.18):

    discounted_price = mrp * (1 - discount)

    final_price = discounted_price * (1 + tax)

    return round(final_price, 2)


print(net_price(1000))

print(net_price(1000, discount=0.20))

print(net_price(1000, discount=0.0, tax=0.05))
```

**Q13.** Write a function `clean_money(text)` that takes a messy price string like `"  ₹ 12,450.00 "` or `"Rs.999"`, strips currency symbols/commas/spaces, and returns a `float`.
Apply it in a loop to `raw_prices = ["₹1,200", "Rs. 3,500", " ₹850 ", "Rs.4,050"]` to create a clean list of floats, and print the **mean** price.

```python
# ✍️ Your answer
# Clean Prices: [1200.0, 3500.0, 850.0, 4050.0]
# Mean Price: 2400.0
```

```python
# ✍️ Your code here
def clean_money(text):

    text = text.replace("₹", "")
    text = text.replace("Rs.", "")
    text = text.replace(",", "")
    text = text.strip()

    return float(text)


raw_prices = ["₹1,200", "Rs. 3,500", " ₹850 ", "Rs.4,050"]

clean_prices = []

for price in raw_prices:
    clean_prices.append(clean_money(price))

mean_price = sum(clean_prices) / len(clean_prices)

print("Clean Prices:", clean_prices)
print("Mean Price:", round(mean_price, 2))
```

**Q14.** Write a function `mean_and_median(numbers)` (using pure Python `sorted()`, `sum()`, and `len()`) that returns `(mean_val, median_val)` for any list of numbers (handle both odd and even lengths).
Test it on `[120, 450, 200, 180, 3000]` and `[10, 20, 30, 40]`.

```python
# ✍️ Your answer
(790.0, 200)
(25.0, 25.0)
```

```python
# ✍️ Your code here
def mean_and_median(numbers):

    mean_val = sum(numbers) / len(numbers)

    sorted_numbers = sorted(numbers)
    n = len(sorted_numbers)

    if n % 2 == 1:
        median_val = sorted_numbers[n // 2]

    else:
        middle1 = sorted_numbers[(n // 2) - 1]
        middle2 = sorted_numbers[n // 2]
        median_val = (middle1 + middle2) / 2

    return mean_val, median_val


print(mean_and_median([120, 450, 200, 180, 3000]))
print(mean_and_median([10, 20, 30, 40]))
```

**Q15.** Write a function `delivery_sla_status(actual_days, promised_days=3)` that returns:
* `"Early"` if `actual_days < promised_days`
* `"On Time"` if `actual_days == promised_days`
* `"Delayed"` if `actual_days > promised_days`

Given `deliveries = [2, 3, 5, 1, 3, 4, 2, 6, 3, 2]`, use your function to count how many orders were `"Early"`, `"On Time"`, and `"Delayed"`.

```python
# ✍️ Your answer
{'Early': 4, 'On Time': 3, 'Delayed': 3}
```

```python
# ✍️ Your code here
def delivery_sla_status(actual_days, promised_days=3):
    if actual_days < promised_days:
        return "Early"
    elif actual_days == promised_days:
        return "On Time"
    else:
        return "Delayed"

# Given deliveries list
deliveries = [2, 3, 5, 1, 3, 4, 2, 6, 3, 2]

# Count each category
counts = {"Early": 0, "On Time": 0, "Delayed": 0}

for days in deliveries:
    status = delivery_sla_status(days)
    counts[status] += 1

print(counts)
```
