**Q1.** Monthly revenue (in ₹ thousands) for 2025 is stored in a 12-element list (Jan to Dec):
`rev = [120, 135, 150, 142, 160, 175, 168, 180, 195, 240, 260, 230]`
Using list indexing and slicing, print:
1. January revenue and December revenue
2. Q1 revenue (first 3 months) and its sum
3. Q4 revenue (last 3 months) and its mean

````python
# ✍️ Your answer
# January Revenue: 120
# December Revenue: 230

# Q1 Revenue: [120, 135, 150]
# Q1 Sum: 405

# Q4 Revenue: [240, 260, 230]
# Q4 Mean: 243.33
````

````python
# ✍️ Your code here
# Monthly revenue (₹ thousands)
rev = [120, 135, 150, 142, 160, 175, 168, 180, 195, 240, 260, 230]

# 1. January and December revenue
print("January Revenue:", rev[0])
print("December Revenue:", rev[-1])

# 2. Q1 revenue (Jan-Mar) and its sum
q1 = rev[:3]
print("\nQ1 Revenue:", q1)
print("Q1 Sum:", sum(q1))

# 3. Q4 revenue (Oct-Dec) and its mean
q4 = rev[-3:]
q4_mean = sum(q4) / len(q4)

print("\nQ4 Revenue:", q4)
print("Q4 Mean:", round(q4_mean, 2))
````

**Q9.** Customer IDs who purchased in January and February are:
`jan_buyers = {"C01", "C02", "C03", "C04", "C05", "C06"}`
`feb_buyers = {"C04", "C05", "C06", "C07", "C08"}`
Using set operations (`&`, `-`, `|`), find and print:
1. **Retained customers** (bought in BOTH Jan and Feb)
2. **Churned customers** (bought in Jan, but NOT in Feb)
3. **New customers in Feb** (bought in Feb, but NOT in Jan)
4. **Total unique customers** across both months

````python
# ✍️ Your answer
# Retained Customers: {'C05', 'C06', 'C04'}
# Churned Customers: {'C03', 'C02', 'C01'}
# New Customers: {'C07', 'C08'}
# Total Unique Customers: {'C07', 'C04', 'C05', 'C06', 'C01', 'C03', 'C02', 'C08'}
# Count of Unique Customers: 8
````

````python
# ✍️ Your code here
# Customer IDs
jan_buyers = {"C01", "C02", "C03", "C04", "C05", "C06"}
feb_buyers = {"C04", "C05", "C06", "C07", "C08"}

# 1. Retained customers (both months)
retained = jan_buyers & feb_buyers

# 2. Churned customers (Jan but not Feb)
churned = jan_buyers - feb_buyers

# 3. New customers in Feb (Feb but not Jan)
new_customers = feb_buyers - jan_buyers

# 4. Total unique customers across both months
total_unique = jan_buyers | feb_buyers

print("Retained Customers:", retained)
print("Churned Customers:", churned)
print("New Customers:", new_customers)
print("Total Unique Customers:", total_unique)
print("Count of Unique Customers:", len(total_unique))
````

**Q10.** Using `jan_buyers` and `feb_buyers` from Q9, calculate the **January-to-February customer retention rate** as a percentage:
`(number of retained customers / number of Jan buyers) * 100`.

````python
# ✍️ Your answer
# Retained Customers: {'C05', 'C06', 'C04'}
# Retention Rate: 50.00%
````

**Q12.** Given a list of payment methods used in 12 orders:
`payments = ["UPI", "Card", "UPI", "COD", "UPI", "Card", "UPI", "UPI", "COD", "Card", "UPI", "Card"]`
Build a frequency dictionary `counts` that counts how many times each payment method appears, and print the **mode** (most frequent payment method).

````python
# ✍️ Your answer
# Frequency Dictionary: {'UPI': 6, 'Card': 4, 'COD': 2}
# Mode: UPI
````

````python
# ✍️ Your code here
# Payment methods used in orders
payments = ["UPI", "Card", "UPI", "COD", "UPI", "Card",
            "UPI", "UPI", "COD", "Card", "UPI", "Card"]

# Build frequency dictionary
counts = {}

for payment in payments:
    counts[payment] = counts.get(payment, 0) + 1

print("Frequency Dictionary:", counts)

# Find the mode (most frequent payment method)
mode = max(counts, key=counts.get)

print("Mode:", mode)
````

**Q13.** Here is a list of dictionaries representing 6 orders (just like rows of a database table):
```python
orders_table = [
    {"order_id": "O1", "region": "North", "category": "Electronics", "amount": 25000, "status": "Delivered"},
    {"order_id": "O2", "region": "South", "category": "Apparel",     "amount": 3200,  "status": "Delivered"},
    {"order_id": "O3", "region": "North", "category": "Furniture",   "amount": 14000, "status": "Cancelled"},
    {"order_id": "O4", "region": "West",  "category": "Electronics", "amount": 18000, "status": "Delivered"},
    {"order_id": "O5", "region": "South", "category": "Electronics", "amount": 32000, "status": "Delivered"},
    {"order_id": "O6", "region": "North", "category": "Apparel",     "amount": 4500,  "status": "Delivered"},
]
```
Filter for **only `"Delivered"` orders**, and calculate their **mean** and **median** `amount`.

````python
# ✍️ Your answer
# Delivered Amounts: [25000, 3200, 18000, 32000, 4500]
# Mean Amount: 16540.00
# Median Amount: 18000.00
````

````python
# ✍️ Your code here
# Orders table
orders_table = [
    {"order_id": "O1", "region": "North", "category": "Electronics", "amount": 25000, "status": "Delivered"},
    {"order_id": "O2", "region": "South", "category": "Apparel",     "amount": 3200,  "status": "Delivered"},
    {"order_id": "O3", "region": "North", "category": "Furniture",   "amount": 14000, "status": "Cancelled"},
    {"order_id": "O4", "region": "West",  "category": "Electronics", "amount": 18000, "status": "Delivered"},
    {"order_id": "O5", "region": "South", "category": "Electronics", "amount": 32000, "status": "Delivered"},
    {"order_id": "O6", "region": "North", "category": "Apparel",     "amount": 4500,  "status": "Delivered"},
]

# Filter Delivered orders
delivered_orders = [
    order for order in orders_table
    if order["status"] == "Delivered"
]

# Extract amounts
amounts = [order["amount"] for order in delivered_orders]

# Mean
mean_amount = sum(amounts) / len(amounts)

# Median
sorted_amounts = sorted(amounts)
median_amount = sorted_amounts[len(sorted_amounts) // 2]

print("Delivered Amounts:", amounts)
print(f"Mean Amount: {mean_amount:.2f}")
print(f"Median Amount: {median_amount:.2f}")
````

**Q14.** Using `orders_table` from Q13, aggregate total `"Delivered"` revenue **by `region`** into a dictionary `region_rev` (e.g., `{"North": ..., "South": ..., "West": ...}`) and print it.

````python
# ✍️ Your answer
{'North': 29500, 'South': 35200, 'West': 18000}
````

````python
# ✍️ Your code here
# Orders table
orders_table = [
    {"order_id": "O1", "region": "North", "category": "Electronics", "amount": 25000, "status": "Delivered"},
    {"order_id": "O2", "region": "South", "category": "Apparel",     "amount": 3200,  "status": "Delivered"},
    {"order_id": "O3", "region": "North", "category": "Furniture",   "amount": 14000, "status": "Cancelled"},
    {"order_id": "O4", "region": "West",  "category": "Electronics", "amount": 18000, "status": "Delivered"},
    {"order_id": "O5", "region": "South", "category": "Electronics", "amount": 32000, "status": "Delivered"},
    {"order_id": "O6", "region": "North", "category": "Apparel",     "amount": 4500,  "status": "Delivered"},
]

# Aggregate delivered revenue by region
region_rev = {}

for order in orders_table:
    if order["status"] == "Delivered":
        region = order["region"]
        amount = order["amount"]

        region_rev[region] = region_rev.get(region, 0) + amount

print(region_rev)
````

**Q15.** Sort the delivered orders from `orders_table` in **descending order of `amount`** using `sorted(..., key=lambda r: r["amount"], reverse=True)` and print the top 2 order IDs with their amounts.

````python
# ✍️ Your answer
# O5 - 32000
# O1 - 25000
````

````python
# ✍️ Your code here
# Delivered orders only
delivered_orders = [
    order for order in orders_table
    if order["status"] == "Delivered"
]

# Sort by amount (highest to lowest)
sorted_orders = sorted(
    delivered_orders,
    key=lambda r: r["amount"],
    reverse=True
)

# Print top 2 orders
for order in sorted_orders[:2]:
    print(order["order_id"], "-", order["amount"])
````
